```python
#Importing Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#To see just the numbers of missing content
df.isnull().sum()
```

```python
#Checking for any missing values via a diagram
df.isnull().sum().plot(kind="bar", figsize=(8,4))
# if we have missing data in column it will be shown in the bar chart, if there is no missing data, the bar chart will be empty.
```

```python
#Data Cleaning
#Dropping all the missing values in rows
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
#Checking duplicated values
df.duplicated().sum()
```

```python
#another way to do it.
df[df.duplicated()]
```

```python
#Deleting all duplicated records 
df.drop_duplicates(inplace=True)
df.duplicated().sum()
#df[df.duplicated()] Would return 0 rows because all duplicates have been removed
```

```python
#Now to check it again with the help of a diagram
df.isnull().sum().plot(kind="bar", figsize=(8,4))
```

```python
df.dtypes
```

```python
#Changing data type of id, host id to object, ie type casting here
df["id"] = df["id"].astype(object)

df["host_id"] = df["host_id"].astype(object)
```

```python
df.dtypes #New ones
```

```python
#Identifying the outliers first

df =df[df["price"] < 1500]

sns.boxplot(data=df, x="price")
```

```python
#Data Analysis
#For price distribution
plt.figure(figsize=(8,6))
sns.histplot(data=df, x='price', kde=True, bins=100)
plt.title("Price Distribution")
plt.ylabel("Frequency")
plt.show
```

```python
df.dtypes
```

```python
#For availability
plt.figure(figsize=(6,4))
sns.histplot(data=df, x='availability_365')
plt.title("availability_365")
plt.ylabel("Frequency")
plt.show
```

```python
df.groupby(by="neighbourhood_group")["price"].mean()
```

```python
#Adding a new column,  price per bed
df["price_per_bed"] = df["price"]/df["beds"]
```

```python
df.groupby(by="neighbourhood_group")["price_per_bed"].mean() #Average price per bed by neighbourhood group
```

```python
df.head() #Can see a new column named price_by_bed
```

```python
sns.barplot(data = df, x="neighbourhood_group", y="price", hue="room_type")#New graphs
```

```python
#For number of reviews and price relation 
plt.figure(figsize=(8,6))
plt.title("Review and Locality Dependency")
sns.scatterplot(data = df, x="number_of_reviews", y="price", hue="neighbourhood_group")
```
